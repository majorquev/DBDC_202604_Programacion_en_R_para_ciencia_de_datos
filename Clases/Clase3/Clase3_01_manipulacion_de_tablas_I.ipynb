### 4.4 Función mutate()

Una de las funciones más útiles de dplyr es la función mutate(), la cual permite crear nuevas columnas, o modificar las existentes.

```
# mutate() ----------------------------------------------------------------
# Crear una nueva columna llamada retraso_neto, que contenga la suma de arr_delay y dep_delay

#antes
vuelos$retraso_neto <- vuelos$arr_delay + vuelos$dep_delay # altera directamente el dataframe
#ahora
mutate(vuelos, retraso_neto2 = arr_delay + dep_delay)
```

```
# Recordar que esta nueva columna solo existe en la ejecución y no se ha modificado el dataset original.
head(vuelos)
vuelos <- mutate(vuelos, retraso_neto2 = arr_delay + dep_delay)
```

```
head(vuelos)
```

```
head(vuelos)
```

```
# Primero defina una nueva data llamada muestra_fl que contenga la fecha, los retrasos, la distancia, y el tiempo de vuelo
muestra_fl <- select(vuelos,anio,mes,dia, arr_delay, dep_delay, distance, air_time)
head(muestra_fl)

```

```
# Cree una columna llamada gain que contenga la diferenecia entre arr_delay y dep_delay, y otra llamada speed que mida la velocidad en horas (distancia / tiempo de vuelo * 60)
mutate(muestra_fl,
       gain = arr_delay - dep_delay,
       speed = 60*distance/air_time,
       gain_abs = abs(gain),
       arr_delay = arr_delay -5) # Podemos modificar columnas existentes
# NOTA: No hemos modificado aun el dataset original (muestra_fl)
```

```
head(muestra_fl)
#si queremos que permanezcan ls cambios, redefinimos con "<-"
muestra_fl <- mutate(muestra_fl,
                        gain = arr_delay - dep_delay,
                        speed = 60*distance/air_time,
                        gain_abs = abs(gain),
                        arr_delay = arr_delay -5)
```

### Operador `%>%` o  `|>`

El operador estrella de dplyr. Funciona como una composición de funciones, y nos permite generar _pipe-lines_, permitiendo  que la generación  del código sea más ordenada, y con ello, su lectura más sencilla.

Este operdaor actua de la sifuiente manera, <br>

    x %>% f(y) , es equivalente a,  f(x,y)



```
# Operador %>% ----------------------------------------------------------------
# Ejemplo sencillo

# head(vuelos)


# Utilizando %>%
vuelos %>% head(3)

```

```
# Genere el conteo de vuelos por mes y día,
# luego filtre por aquellos registros donde el día es 1,
# ordene de manera descendente por n,
# finalmente muestra la cabecera del dataset resultante.
# NUEVO: antes de filtrar, luego del count, agregar una columna que repita el total de vuelos de todo el dataframe, llamada "total" (suma de todos los n)
## ANTES
head(arrange(filter(mutate(count(vuelos,mes,dia, name = "conteo"), total_vuelos = sum(conteo)), dia == 1), desc(conteo)))
```

```
?count
```

```
# Genere el conteo de vuelos por mes y día,
# luego filtre por aquellos registros donde el día es 1,
# ordene de manera descendente por n,
# finalmente muestra la cabecera del dataset resultante.
# NUEVO: antes de filtrar, luego del count, agregar una columna que repita el total de vuelos de todo el dataframe, llamada "total" (suma de todos los n)
## AHORA
vuelos |>
  count(mes,dia, name = "conteo") |>
  mutate(total_vuelos = sum(conteo)) |>
  filter(dia == 1) |>
  arrange(desc(conteo)) |>
  head()
```

```
# A la onsulta anterior podríamos agregar una nueva columna indicando con un "A" si el conteo vuelos es mayor a 970, y un "B" si no.
vuelos |>
  count(mes,dia, name = "conteo") |>
  mutate(total_vuelos = sum(conteo),
         cat_vuelos = ifelse(conteo > 970, "A","B") ) |>
  filter(dia == 1) |>
  arrange(desc(conteo)) |>
  head()
```

```
#Alternativamente: case_when()
# Incluir el nombre del mes. Si mes = 1 , colocar "ene", es 2, colocar "feb",....
vuelos |>
  count(mes,dia, name = "conteo") |>
  mutate(total_vuelos = sum(conteo),
         cat_vuelos = ifelse(conteo > 970, "A","B") ,
         mes_str = case_when(mes == 1 ~ "ene",
                             mes == 2 ~ "feb",
                             mes == 3 ~ "mar",
                             mes == 4 ~ "abr",
                             mes == 5 ~ "may",
                             mes == 6 ~ "jun",
                             mes == 7 ~ "jul",
                             mes == 8 ~ "ago",
                             mes == 9 ~ "sep",
                             mes == 10 ~ "oct",
                             mes == 11 ~ "nov",
                             TRUE ~ "dic" ) ) |>
  filter(dia == 1) |>
  arrange(desc(conteo)) |>
  head()
```

### 4.5 Función group_by() y summarise()

La función `summarise()` calcula medidas agregadas sobre columnas, mientras que la función `group_by`, genera un data.frame "en bloques" dados por la(s) columna(s) ingresadas como parámetros en la función.



```
# group_by() y summarise() ----------------------------------------------------------------
# Ejemplo sencillo
exam2 <- data.frame(
  x = c(1:5),
  y = c(7:11),
  z = c("A", "B", "A", "B", "C")
)
exam2
```

```
group_by(exam2, z)
# exam2 %>% group_by(z)
```

```
exam2 %>%
  mutate(tot_y = sum(y))
```

```

 exam2 %>%
  group_by(z) %>%
  mutate(tot_y = sum(y))
```

```
# Podemos generar un número de fila por cada grupo
 exam2 %>%
   mutate(rn = row_number())

exam2 %>%
   group_by(z) %>%
   mutate(rn = row_number())
```

```
# 1) Aplicar la función summarise a la columna 'y' y calcular su suma, la suma de x y el promedio de x
exam2 %>%
  summarise(tot_y = sum(y),
            tot_x = sum(x),
            mean_x = mean(x))

# 2) Generar grupos según la variable z. (a diferencia de sql, el comando group_by() no requiere de una medida agregada para que este funcione.)
exam2 %>%
  group_by(z) %>%
  summarise(tot_y = sum(y),
            tot_x = sum(x),
            mean_x = mean(x))

```

```
# 3) Calcular media de x e y en cada valor de z.
# podemos calcular máximos y mínimos
exam2 %>%
  group_by(z) %>%
  summarise(mean_y = mean(y),
            max_x = max(x),
            mean_x = mean(x))
```
